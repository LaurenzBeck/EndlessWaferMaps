```python
from pathlib import Path

from skald import combine_runs
import seaborn as sns
import polars as pl

import matplotlib.pyplot as plt
from matplotlib.ticker import LogLocator
import numpy as np

from experiments.visualizations import set_paper_style

set_paper_style()
```

```python
metrics = (
    combine_runs(
        Path(r"W:\EndlessWM\experiments\reports\parameter_coverage"),
        include_params=["data.name", "data.param_fraction", "data.dataset_seed"],
    )
    .rename({"data.name": "dataset", "data.param_fraction": "parameter space coverage"})
    .filter(
        pl.col("epoch").is_null(),
        name="mAP",
        pattern="*",
        task=0,
    )
)
metrics
```

```python
metrics["stream"].unique()
```

```python
facet_grid = sns.relplot(
    data=metrics.filter(pl.col("epoch").is_null()).to_pandas(),
    x="parameter space coverage",
    y="value",
    hue="stream",
    kind="line",
    col="dataset",
    col_wrap=2,
    col_order=["i-1k", "ii-1k", "iii-1k", "iv-1k"],
)

for ax in facet_grid.axes.flatten():
    ax.set_xlabel("parameter space coverage", fontsize=9)
    ax.set_ylabel("mAP", fontsize=9)
    ax.tick_params(labelsize=8)

facet_grid
```

```python
dataset_order = [
    "I",
    "II",
    "III",
    "IV",
]

fig, ax = plt.subplots(figsize=(3.5, 2.25))

ax = sns.lineplot(
    data=metrics.filter(
        pl.col("epoch").is_null(),
    )
    .with_columns(pl.col("dataset").replace({"i-1k": "I", "ii-1k": "II", "iii-1k": "III", "iv-1k": "IV"}))
    .filter(pl.col("stream") == pl.col("dataset")),
    x="parameter space coverage",
    y="value",
    hue="dataset",
    hue_order=dataset_order,
    ax=ax,
)

ax.set(
    xlabel="parameter space coverage",
    ylabel="mAP",
)

sns.move_legend(
    ax,
    "lower center",
    bbox_to_anchor=(0.5, 1.02),
    ncol=len(dataset_order),
    title=None,
    frameon=True,
    borderaxespad=1.0,
    borderpad=0.25,
    handlelength=1.0,
    handletextpad=0.5,
    columnspacing=1.0,
    labelspacing=0.2,
)

fig.tight_layout()

fig
```
